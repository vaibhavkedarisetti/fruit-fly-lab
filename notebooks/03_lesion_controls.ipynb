# Looming lesion controls

This notebook tests a falsifiable pathway claim. It compares looming, receding, static, and looming conditions with LC4 and/or LPLC2 silenced. Silencing removes outgoing contribution from the specified real FlyWire neurons; it does not change topology.

```python
import importlib
import pandas as pd
from brain.neurons.registry import load_connectome
from brain.sensory.retinotopy import load_retinotopy

controls = importlib.import_module('experiments.02_escape_controls')
connectome = load_connectome()
retinotopy = load_retinotopy(connectome)
conditions = [('looming', ()), ('receding', ()), ('static', ()), ('looming', ('LC4',)), ('looming', ('LPLC2',)), ('looming', ('LC4', 'LPLC2'))]
results = [controls.run_condition(connectome, retinotopy, kind, silenced, seed=0) for kind, silenced in conditions]
pd.DataFrame(results)[['condition', 'dnp01_spikes', 'peak_dnp01_hz', 'neurons_active', 'total_spikes']]
```

## Interpretation

The expected control is that the LC4+LPLC2 lesion removes the DNp01 response while the stimulus remains present. If it does not, report the result rather than tuning the model.
