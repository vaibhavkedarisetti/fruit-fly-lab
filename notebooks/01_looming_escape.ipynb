# Looming → escape

This tutorial runs the supported visual looming experiment. It drives real LC4 and LPLC2 FlyWire populations through the documented encoder, then reads the real DNp01 (Giant Fibre) population.

Run Jupyter from the repository root after building the v783 connectome.

```python
import importlib
import pandas as pd

looming = importlib.import_module('experiments.01_looming_escape')
session, stimulus, rows = looming.run(seed=0, quiet=True)
table = pd.DataFrame(rows)
table.tail()
```

```python
dnp01 = session.c.by_cell_type('DNp01')['idx'].to_numpy()
summary = {
    'dataset': session.c.dataset,
    'DNp01 total spikes': int(session.engine.spike_counts[dnp01].sum()),
    'active neurons': int((session.engine.spike_counts > 0).sum()),
    'total network spikes': int(session.engine.spike_counts.sum()),
}
summary
```
